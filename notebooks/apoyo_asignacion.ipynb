```r
# Instalamos las librerías si no están presentes
if (!require("readxl")) install.packages("readxl") # Para guardar excel con datos
if (!require("writexl")) install.packages("writexl") # Para escribir el excel

# Cargar librerias necesarias
library(tidyverse)
library(readxl)
library(writexl)

if (!require("fdth")) install.packages("fdth") # Para procesar las tablas de frecuencia
library(fdth)

if (!require("pdfetch")) install.packages("pdfetch") # Para procesar los datos
library(pdfetch)

```

```r
datos_financieros_nvidia <- pdfetch_YAHOO("NVDA",
                                          from = "2023-01-01",
                                          to   = "2025-12-31",
                                          interval='monthly',
                                          fields= c("open", "volume")) 
```

```r
datos_financieros_nvidia
```

```r
autoplot(datos_financieros_nvidia) +
  geom_line(color = "#c80e0e") +
  labs(title = "Serie de Tiempo Valores Acciones",
       x = "Fechas mensuales",
       y = "Valor")
```

```r
hist(datos_financieros_nvidia$NVDA.volume)
```

```r
ggplot(datos_financieros_nvidia)+
    geom_boxplot(aes(y=NVDA.volume ))+
  labs(title = "Serie de Tiempo Valores Acciones",
       x = "Fechas mensuales",
       y = "Valor")
```

```r
ggplot(datos_financieros_nvidia)+
 geom_histogram( aes(x=NVDA.open ), bins=5, color='orange',fill='purple')
```

```r
apple_incide_simple <- datos_financieros_aaple %>% # pipe operator. (operador de encadenamiento)
    as.data.frame()%>%
    mutate(indice_simple= (AAPL/ first(AAPL))*100)
```

```r
datos_indice_compuesto <- datos_financieros%>%
    as.data.frame()%>%
    mutate(indice_agregado= (AAPL+ IBM )/(first(AAPL)+first(IBM) )*100 )%>% ## CAMBIARLO POR NOMBRE DE LAS ACCIONES SELECCIONADAS
    rownames_to_column(var = "Fecha")%>%
    mutate(Fecha= as.Date(Fecha))
```

```r
datos_indice_compuesto
```

```r
ggplot(datos_indice_compuesto)+
 geom_line(aes(x= Fecha, y= indice_agregado ))
```
